```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
# Load our input image
image = cv2.imread('tiger.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB for displaying with matplotlib

# Use cv2.split to get each color space separately
R,G,B = cv2.split(image)
print(B.shape)
print(G.shape)
print(R.shape)
```

```python
plt.imshow(image)
plt.title('Original Image')
plt.axis('off')
```

```python
plt.imshow(B)
plt.title('Blue Channel')
plt.axis('off')
plt.show()
```

```python
plt.imshow(G)
plt.title('Green Channel')
plt.axis('off')
plt.show()
```

```python
plt.imshow(R)
plt.title('Red Channel')
plt.axis('off')
plt.show()
```

```python
R,G,B = cv2.split(image)

# Let's re-make the original image, 
merged = cv2.merge([R, G, B]) 
plt.imshow(merged)
plt.title("Merged")
plt.axis("off")
plt.show()
```

```python
## HSV Color Space
image = cv2.imread('pinkSky.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB for displaying with matplotlib

# Convert to HSV
hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
plt.subplots(1, 2, figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(hsv_image)
plt.title('HSV')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(image)
plt.title('Original Image')
plt.axis('off')
plt.show()
```

```python
plt.imshow(hsv_image[:, :, 0])
plt.title('Hue')
plt.axis('off')
plt.show()
```

```python
plt.imshow(hsv_image[:, :, 1])
plt.title('Saturation')
plt.axis('off')
plt.show()
```

```python
plt.imshow(hsv_image[:, :, 2])
plt.title('Value')
plt.axis('off')
plt.show()
```
