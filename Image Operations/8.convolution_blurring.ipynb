```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

Convolution in OpenCV is a mathematical operation used to process an image by moving a small matrix called a kernel (or filter) across the image and calculating a new value for each pixel.

```python
image = cv2.imread('tiger.jpg')
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.show()

# Creating our 3 x 3 kernel
kernel_3x3 = np.ones((3, 3), np.float32) / 9

# We use the cv2.fitler2D to conovlve the kernal with an image 
blurred = cv2.filter2D(image, -1, kernel_3x3)
plt.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB))
plt.title('3x3 Kernel Blurring')
plt.show()

# Creating our 7 x 7 kernel
kernel_7x7 = np.ones((7, 7), np.float32) / 49

blurred2 = cv2.filter2D(image, -1, kernel_7x7)
plt.imshow(cv2.cvtColor(blurred2, cv2.COLOR_BGR2RGB))
plt.title('7x7 Kernel Blurring')
plt.show()
```

## Other commonly used blurring methods in OpenCV
Regular Blurring

Gaussian Blurring

Median Blurring

```python
image = cv2.imread('tiger.jpg')

# Averaging done by convolving the image with a normalized box filter. 
# This takes the pixels under the box and replaces the central element
# Box size needs to odd and positive 
blur = cv2.blur(image, (8,8)) # cv2.blur(src, ksize)
plt.imshow(cv2.cvtColor(blur, cv2.COLOR_BGR2RGB))
plt.title('Averaging Blurring')
plt.show()
# Instead of box filter, gaussian kernel 
Gaussian = cv2.GaussianBlur(image, (5,5), 0) # cv2.GaussianBlur(src, ksize, sigmaX)
plt.imshow(cv2.cvtColor(Gaussian, cv2.COLOR_BGR2RGB))
plt.title('Gaussian Blurring')
plt.show()

# Takes median of all the pixels under kernel area and central 
# element is replaced with this median value
median = cv2.medianBlur(image, 5) # cv2.medianBlur(src, ksize)
plt.imshow(cv2.cvtColor(median, cv2.COLOR_BGR2RGB))
plt.title('Median Blurring')
plt.show()
```

```python
# Bilateral is very effective in noise removal while keeping edges sharp
# Unlike regular/Gaussian blur, bilateral filtering tries to smooth noise while preserving edges.
bilateral = cv2.bilateralFilter(image, 9, 75, 75)
plt.imshow(cv2.cvtColor(bilateral, cv2.COLOR_BGR2RGB))
plt.title('Bilateral Blurring')
plt.show()
```

## Image De-noising - Non-Local Means Denoising
There are 4 variations of Non-Local Means Denoising:

*cv2.fastNlMeansDenoising() - works with a single grayscale images
*cv2.fastNlMeansDenoisingColored() - works with a color image.
*cv2.fastNlMeansDenoisingMulti() - works with image sequence captured in short period of time (grayscale images)
*cv2.fastNlMeansDenoisingColoredMulti() - same as above, but for color images.

```python
image = cv2.imread('mountain.jpg')
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.show()

dst = cv2.fastNlMeansDenoisingColored(image, None, 6, 6, 7, 21)
plt.imshow(cv2.cvtColor(dst, cv2.COLOR_BGR2RGB))
plt.title('fastNlMeansDenoisingColored')
plt.show()
'''src	--> Input color image
None -->	Destination; OpenCV creates it automatically
h -->	Strength of luminance/noise filtering
hColor -->	Strength of color-noise filtering
templateWindowSize -->	Size of the window used to compare image patches
searchWindowSize -->	Size of the area searched for similar patches
'''
```
