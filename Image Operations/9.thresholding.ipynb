```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

![alt text](image.png) 

```python
image = cv2.imread('book_page.png',0)
plt.imshow(image, cmap='gray')
plt.title('Original')
plt.axis('off')
plt.show()

# Values below 127 goes to 0 or black, everything above goes to 255 (white)
ret,thresh1 = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)
plt.imshow(thresh1, cmap='gray')
plt.title('1 Threshold Binary @ 127')
plt.axis('off')
plt.show()

# Values below 127 go to 255 and values above 127 go to 0 (reverse of above)
ret,thresh2 = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY_INV)
plt.imshow(thresh2, cmap='gray')
plt.title('2 Threshold Binary Inverse @ 127')
plt.axis('off')
plt.show()

# Values above 127 are truncated (held) at 127 (the 255 argument is unused)
ret,thresh3 = cv2.threshold(image, 127, 255, cv2.THRESH_TRUNC)
plt.imshow(thresh3, cmap='gray')
plt.title('3 THRESH TRUNC @ 127')
plt.axis('off')
plt.show()

# Values below 127 go to 0, above 127 are unchanged  
ret,thresh4 = cv2.threshold(image, 127, 255, cv2.THRESH_TOZERO)
plt.imshow(thresh4, cmap='gray')
plt.title('4 THRESH TOZERO @ 127')
plt.axis('off')
plt.show()

# Reverse of the above, below 127 is unchanged, above 127 goes to 0
ret,thresh5 = cv2.threshold(image, 127, 255, cv2.THRESH_TOZERO_INV)
plt.imshow(thresh5, cmap='gray')
plt.title('5 THRESH TOZERO INV @ 127')
plt.axis('off')
plt.show()
```

```python
image = cv2.imread('book_page.png',0)
plt.imshow(image, cmap='gray')
plt.title('Original')
plt.axis('off')
plt.show()

# Values below 127 goes to 0 (black, everything above goes to 255 (white)
ret,thresh1 = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)
plt.imshow(thresh1, cmap='gray')
plt.title('Threshold Binary')
plt.axis('off')
plt.show()

# It's good practice to blur images as it removes noise
#image = cv2.GaussianBlur(image, (3, 3), 0)

# Using adaptiveThreshold
thresh = cv2.adaptiveThreshold(image, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 3, 5) 
plt.imshow(thresh, cmap='gray')
plt.title('Adaptive Mean Thresholding')
plt.axis('off')
plt.show()

_, th2 = cv2.threshold(image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
plt.imshow(th2, cmap='gray')
plt.title("Otsu's Thresholding")
plt.axis('off')
plt.show()

# Otsu's thresholding after Gaussian filtering
blur = cv2.GaussianBlur(image, (5,5), 0)
_, th3 = cv2.threshold(blur, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
plt.imshow(th3, cmap='gray')
plt.title("Otsu's Thresholding after Gaussian filtering")
plt.axis('off')
plt.show()
```
