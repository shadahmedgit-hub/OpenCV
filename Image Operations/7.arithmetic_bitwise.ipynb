Arithmetic Operations

These are simple operations that allow us to directly add or subract to the color intensity.

Calculates the per-element operation of two arrays. The overall effect is increasing or decreasing brightness.

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
image = cv2.imread('pinkSky.jpg', 0)
image = cv2.resize(image,(500,500))
plt.imshow(image, cmap='gray')
```

```python
M = np.ones(image.shape, dtype = "uint8") * 100 
M.shape
```

## Increasing Brightness

```python
# We use this to add this matrix M, to our image
# Notice the increase in brightness
added = cv2.add(image, M)
plt.imshow(added, cmap='gray')
plt.title("cv2.add() Result")
plt.show()

# Now if we just added it, look what happens
added2 = image + M 
plt.imshow(added2, cmap='gray')
plt.title("Simple Numpy Adding Results in Clipping")
plt.show()
```

## Decreasing Brightness

```python
subtracted = cv2.subtract(image, M)
plt.imshow(subtracted, cmap='gray')
plt.title("cv2.subtract() Result")
plt.show()

subtracted = image - M 
plt.imshow(subtracted, cmap='gray')
plt.title("Simple Numpy Subtraction")
plt.show()
```

## Bitwise Operations and Masking

```python
# Making a square
square = np.zeros((400, 400), np.uint8)
cv2.rectangle(square, (50, 50), (250, 250), 255, -1)
plt.imshow(square, cmap='gray')
plt.title("Square")
plt.show()

# Making a ellipse
ellipse = np.zeros((400, 400), np.uint8)
cv2.ellipse(ellipse, (150, 150), (150, 150), 30, 0, 180, 255, -1)
plt.imshow(ellipse, cmap='gray')
plt.title("Ellipse")
plt.show()
```

```python
# Shows only where they intersect
And = cv2.bitwise_and(square, ellipse)
plt.imshow(And, cmap='gray')
plt.title("bitwise_and() Result")
plt.show()

# Shows where either square or ellipse is 
bitwiseOr = cv2.bitwise_or(square, ellipse)
plt.imshow(bitwiseOr, cmap='gray')
plt.title("bitwise_or() Result")
plt.show()

# Shows where either exist by itself
bitwiseXor = cv2.bitwise_xor(square, ellipse)
plt.imshow(bitwiseXor, cmap='gray')
plt.title("bitwise_xor() Result")
plt.show()

# Shows everything that isn't part of the square
bitwiseNot_sq = cv2.bitwise_not(square)
plt.imshow(bitwiseNot_sq, cmap='gray')
plt.title("bitwise_not() Result")
plt.show()
```
