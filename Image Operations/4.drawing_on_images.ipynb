````python
import cv2
import numpy as np
from matplotlib import pyplot as plt
````

````python
# Create a black image using numpy to create and array of black 
image = np.zeros((512,512,3), np.uint8)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Can we make this in black and white? grayscale
image_gray = np.zeros((512,512), np.uint8)

# Black would be the same as a greyscale or color image (same for white)
plt.imshow(image_rgb)
plt.title("RGB Image")
plt.axis("off")
plt.show()
plt.imshow(image_gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

plt.show()
````

````python
cv2.line(image_rgb, (0,0), (511,511), (255,127,0), 5)

plt.imshow(image_rgb)
plt.title("Black Canvas With Diagonal Line")
plt.axis("off")
plt.show()
````

````python
cv2.line(image_gray, (0,0), (511,511), (255,127,0), 5)

plt.imshow(image_gray,cmap='gray')
plt.title("Black Canvas With Diagonal Line")
plt.axis("off")
plt.show()
````

````python
image = np.zeros((512,512,3), np.uint8)

# Thickness - if positive. Negative thickness means that it is filled
cv2.rectangle(image, (100,100), (300,400), (127,50,127), 10)
plt.imshow(image)
plt.title("Black Canvas With Pink Rectangle")
# plt.axis("off")
plt.show()
````

cv2.cirlce(image, center, radius, color, fill)

````python
image = np.zeros((512,512,3), np.uint8)

cv2.circle(image, (350, 350), 50, (15,150,50), -1) # -1 means filled
plt.imshow(image)
plt.title("Black Canvas With Green Circle")
plt.show()
````

Polygons

cv2.polylines(image, points, Closed?, color, thickness)

if Closed = True, we join the first and last points.

````python
image = np.zeros((512,512,3), np.uint8)

# Let's define four points
pts = np.array( [[10,50], [400,50], [90,200], [50,500]], np.int32)

# Let's now reshape our points in form  required by polylines
pts = pts.reshape((-1,1,2))

cv2.polylines(image, [pts], True, (0,0,255), 3)
plt.imshow(image)
plt.title("Black Canvas with Red Polygon")
plt.axis("off")
plt.show()
````

### **And now for adding text with cv2.putText**

```cv2.putText(image, 'Text to Display', bottom left starting point, Font, Font Size, Color, Thickness)```

**Available Fonts**
- FONT_HERSHEY_SIMPLEX, FONT_HERSHEY_PLAIN
- FONT_HERSHEY_DUPLEX,FONT_HERSHEY_COMPLEX 
- FONT_HERSHEY_TRIPLEX, FONT_HERSHEY_COMPLEX_SMALL
- FONT_HERSHEY_SCRIPT_SIMPLEX
- FONT_HERSHEY_SCRIPT_COMPLEX

````python
image = np.zeros((1000, 1000, 3), np.uint8)

ourString = 'Hello World!'

cv2.putText(image, ourString, (155, 290),
            cv2.FONT_HERSHEY_SCRIPT_COMPLEX, 3, (40, 200, 0), 4)

cv2.putText(image, ourString, (155, 490),
            cv2.FONT_HERSHEY_COMPLEX_SMALL, 3, (40, 200, 0), 4)

cv2.putText(image, ourString, (155, 690),
            cv2.FONT_HERSHEY_DUPLEX, 3, (40, 200, 0), 4)

cv2.putText(image, ourString, (155, 890),
            cv2.FONT_HERSHEY_SCRIPT_COMPLEX, 3, (40, 200, 0), 4)

plt.imshow(image)
plt.title("Different OpenCV Font Styles")
plt.axis("off")
plt.show()
````
